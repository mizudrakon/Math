## Euclidean algorithm
- works with 2 numbers $a$ and $b$ that are both $\in \mathbb{Z}$
- they're subtracted from each other in turns to get a sort of common reminder
- $a - b = r_0$, $b - r = r_1$, ... 
- the final reminder $r_n$ is guaranteed to be **the greatest common divisor**

```python
# gcd algorithm keeping a record of a % b = r
def gcd(a,b,record):
    if b > a:
        a,b = b,a
    r = a % b
    record += [[a,b,r]]
    while r != 0:
        a,b = b,r
        r = a % b
        record += [[a,b,r]]
    return b
```
