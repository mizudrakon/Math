### using LaTeX code for vectors and matrices
Here we have a "column" vector: $\begin{pmatrix}
x \\
y \\
z
\end{pmatrix}$, and here we have a "row" vector: $\begin{pmatrix}x & y & z\end{pmatrix}$

$\mathbf{a} \cdot \mathbf{b} = \sum_{i=1}^n a_i b_i$, 

vector: $\mathbf{v}=\begin{pmatrix}1 & 2 & 3\end{pmatrix}$, $v_1 = 1$

$\begin{vmatrix}
    a_1 & a_2 & a_3 \\
    b_1 & b_2 & b_3 \\
    c_1 & c_2 & c_3
\end{vmatrix} = \begin{vmatrix}
    \mathbf{v_1} \\
    \mathbf{v_2} \\
    \mathbf{v_3}
\end{vmatrix}$ 

### unit vectors
$\mathbf{v} = v_x\hat{i} + v_y\hat{j} + v_z\hat{k}$

### Norm
$\|\mathbf{v}\| = \sqrt{v_1^2 + v_2^2 + v_3^2}$
